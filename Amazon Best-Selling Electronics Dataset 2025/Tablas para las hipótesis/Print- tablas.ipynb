```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import glob

ruta_base = r"C:\Users\juanb\Desktop\IronHack\Proyecto individual\proyecto-analysis-gaming\Amazon Best-Selling Electronics Dataset 2025\Tablas para las hipótesis"
ruta_output = os.path.join(ruta_base, "Output proyecto")

os.makedirs(ruta_output, exist_ok=True)

archivos_csv = [
    "hipótesis 1 - tabla 1.csv",
    "hipótesis 2- tabla 2.csv",
    "hipótesis 3- tabla 3.csv"
]

for nombre_csv in archivos_csv:
    ruta_csv = os.path.join(ruta_base, nombre_csv)
    
    if not os.path.exists(ruta_csv):
        print(f"No encontrado: {ruta_csv}")
        continue

    try:
        try:
            df = pd.read_csv(ruta_csv, encoding="utf-8")
        except:
            df = pd.read_csv(ruta_csv, encoding="cp1252")

        fig, ax = plt.subplots(figsize=(14, max(3, 0.5 * (len(df) + 1))))
        ax.axis("off")

        tabla = ax.table(
            cellText=df.values,
            colLabels=df.columns,
            cellLoc="center",
            loc="center"
        )

        tabla.auto_set_font_size(False)
        tabla.set_fontsize(10)
        tabla.scale(1, 1.4)

        nombre_png = nombre_csv.replace(".csv", ".png")
        ruta_png = os.path.join(ruta_output, nombre_png)

        plt.savefig(ruta_png, bbox_inches="tight", dpi=300)
        plt.close()

        print(f"Guardado: {ruta_png}")

    except Exception as e:
        print(f"Error con {nombre_csv}: {e}")
```
