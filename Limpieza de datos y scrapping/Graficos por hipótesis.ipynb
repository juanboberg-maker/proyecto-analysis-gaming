```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Configuración de estilo profesional
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 8)
plt.rcParams['font.size'] = 11
plt.rcParams['axes.labelsize'] = 12
plt.rcParams['axes.titlesize'] = 14
plt.rcParams['xtick.labelsize'] = 10
plt.rcParams['ytick.labelsize'] = 10
plt.rcParams['legend.fontsize'] = 10

# ===== RUTAS (MODIFICADAS) =====
h1_dir = Path(r"C:\Users\juanb\Desktop\IronHack\Proyecto individual\proyecto-analysis-gaming\Tablas\H1")
h2_dir = Path(r"C:\Users\juanb\Desktop\IronHack\Proyecto individual\proyecto-analysis-gaming\Tablas\H2")
h3_dir = Path(r"C:\Users\juanb\Desktop\IronHack\Proyecto individual\proyecto-analysis-gaming\Tablas\H3")

# ===== CARGAR DATOS =====
df_h1 = pd.read_csv(h1_dir / "h1_dataset_completo.csv")
df_h2 = pd.read_csv(h2_dir / "h2_dataset_completo.csv")
df_h3 = pd.read_csv(h3_dir / "h3_dataset_completo.csv")

print("="*60)
print("GENERANDO GRÁFICOS PARA LAS 3 HIPÓTESIS")
print("="*60)

# ============================================================
# H1: GRÁFICOS (Guardar en carpeta H1)
# ============================================================

print("\n📊 Generando gráficos para H1...")

# ===== H1.1: Rating Top 10% vs Resto (Global) =====
fig, ax = plt.subplots(figsize=(10, 6))

top10 = df_h1[df_h1['top_10_popularity'] == True]
resto = df_h1[df_h1['top_10_popularity'] == False]

data_h1_global = pd.DataFrame({
    'Grupo': ['Top 10%\nPopularidad', 'Resto'],
    'Rating': [top10['rating'].mean(), resto['rating'].mean()],
    'N': [len(top10), len(resto)]
})

bars = ax.bar(data_h1_global['Grupo'], data_h1_global['Rating'], 
               color=['#2ecc71', '#95a5a6'], edgecolor='black', linewidth=1.5)

for i, (bar, row) in enumerate(zip(bars, data_h1_global.itertuples())):
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height,
            f'{row.Rating:.3f}\n(n={row.N})',
            ha='center', va='bottom', fontweight='bold', fontsize=11)

ax.set_ylabel('Rating Promedio', fontweight='bold')
ax.set_title('H1: Rating Promedio - Top 10% Popularidad vs Resto', 
             fontweight='bold', fontsize=14)
ax.set_ylim(4.0, 5.0)
ax.axhline(y=df_h1['rating'].mean(), color='red', linestyle='--', 
           linewidth=2, label=f'Media global: {df_h1["rating"].mean():.3f}')
ax.legend()

plt.tight_layout()
plt.savefig(h1_dir / "h1_rating_global.png", dpi=300, bbox_inches='tight')
print(f"✅ h1_rating_global.png → {h1_dir}")
plt.close()

# ===== H1.2: Rating por Subcategoría (Top 10% vs Resto) =====
fig, ax = plt.subplots(figsize=(14, 7))

subcategorias = sorted(df_h1['subcategory'].unique())
x = np.arange(len(subcategorias))
width = 0.35

ratings_top10 = []
ratings_resto = []

for subcat in subcategorias:
    subcat_data = df_h1[df_h1['subcategory'] == subcat]
    top10_subcat = subcat_data[subcat_data['top_10_popularity'] == True]['rating'].mean()
    resto_subcat = subcat_data[subcat_data['top_10_popularity'] == False]['rating'].mean()
    ratings_top10.append(top10_subcat)
    ratings_resto.append(resto_subcat)

bars1 = ax.bar(x - width/2, ratings_top10, width, label='Top 10%', 
               color='#3498db', edgecolor='black')
bars2 = ax.bar(x + width/2, ratings_resto, width, label='Resto', 
               color='#e74c3c', edgecolor='black')

ax.set_xlabel('Subcategoría', fontweight='bold')
ax.set_ylabel('Rating Promedio', fontweight='bold')
ax.set_title('H1: Rating por Subcategoría - Top 10% vs Resto', 
             fontweight='bold', fontsize=14)
ax.set_xticks(x)
ax.set_xticklabels(subcategorias)
ax.legend()
ax.set_ylim(4.0, 5.0)
ax.axhline(y=df_h1['rating'].mean(), color='gray', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.savefig(h1_dir / "h1_rating_por_subcategoria.png", dpi=300, bbox_inches='tight')
print(f"✅ h1_rating_por_subcategoria.png → {h1_dir}")
plt.close()

# ===== H1.3: Rating por Gama de Precio =====
fig, ax = plt.subplots(figsize=(12, 7))

gamas = ['baja', 'media', 'alta']
x = np.arange(len(gamas))
width = 0.35

ratings_top10_gama = []
ratings_resto_gama = []

for gama in gamas:
    gama_data = df_h1[df_h1['price_tier'] == gama]
    top10_gama = gama_data[gama_data['top_10_popularity'] == True]['rating'].mean()
    resto_gama = gama_data[gama_data['top_10_popularity'] == False]['rating'].mean()
    ratings_top10_gama.append(top10_gama)
    ratings_resto_gama.append(resto_gama)

bars1 = ax.bar(x - width/2, ratings_top10_gama, width, label='Top 10%', 
               color='#2ecc71', edgecolor='black')
bars2 = ax.bar(x + width/2, ratings_resto_gama, width, label='Resto', 
               color='#e67e22', edgecolor='black')

for bars in [bars1, bars2]:
    for bar in bars:
        height = bar.get_height()
        if not np.isnan(height):
            ax.text(bar.get_x() + bar.get_width()/2., height,
                    f'{height:.3f}',
                    ha='center', va='bottom', fontsize=10)

ax.set_xlabel('Gama de Precio', fontweight='bold')
ax.set_ylabel('Rating Promedio', fontweight='bold')
ax.set_title('H1: Rating por Gama de Precio - Top 10% vs Resto', 
             fontweight='bold', fontsize=14)
ax.set_xticks(x)
ax.set_xticklabels([g.upper() for g in gamas])
ax.legend()
ax.set_ylim(4.0, 5.0)

plt.tight_layout()
plt.savefig(h1_dir / "h1_rating_por_gama.png", dpi=300, bbox_inches='tight')
print(f"✅ h1_rating_por_gama.png → {h1_dir}")
plt.close()

# ============================================================
# H2: GRÁFICOS (Guardar en carpeta H2)
# ============================================================

print("\n📊 Generando gráficos para H2...")

# ===== H2.1: Scatter Plot - Rating vs Popularidad por Marca =====
fig, ax = plt.subplots(figsize=(12, 8))

marca_stats = df_h2.groupby('brand').agg(
    n_productos=('product_name', 'count'),
    rating_promedio=('rating', 'mean'),
    popularidad_promedio=('review_count', 'mean'),
).reset_index()

marca_stats = marca_stats[marca_stats['n_productos'] >= 3]

scatter = ax.scatter(marca_stats['rating_promedio'], 
                     marca_stats['popularidad_promedio'],
                     s=marca_stats['n_productos']*50,
                     alpha=0.6, 
                     c=marca_stats['n_productos'],
                     cmap='viridis',
                     edgecolors='black',
                     linewidth=1.5)

for _, row in marca_stats.iterrows():
    ax.annotate(row['brand'], 
                (row['rating_promedio'], row['popularidad_promedio']),
                fontsize=9, alpha=0.8)

z = np.polyfit(marca_stats['rating_promedio'], marca_stats['popularidad_promedio'], 1)
p = np.poly1d(z)
ax.plot(marca_stats['rating_promedio'], p(marca_stats['rating_promedio']), 
        "r--", alpha=0.8, linewidth=2, label=f'Tendencia (r={marca_stats["rating_promedio"].corr(marca_stats["popularidad_promedio"]):.3f})')

ax.set_xlabel('Rating Promedio', fontweight='bold')
ax.set_ylabel('Popularidad Promedio (Opiniones)', fontweight='bold')
ax.set_title('H2: Relación entre Rating y Popularidad por Marca\n(Tamaño = Nº productos)', 
             fontweight='bold', fontsize=14)
ax.legend()
ax.grid(True, alpha=0.3)

cbar = plt.colorbar(scatter, ax=ax)
cbar.set_label('Nº Productos', fontweight='bold')

plt.tight_layout()
plt.savefig(h2_dir / "h2_rating_vs_popularidad.png", dpi=300, bbox_inches='tight')
print(f"✅ h2_rating_vs_popularidad.png → {h2_dir}")
plt.close()

# ===== H2.2: Top 5 Marcas Populares vs Mejor Valoradas =====
# ===== H2.2: Top 5 Marcas Populares vs Mejor Valoradas =====
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))

pct_top10_brand = (
    df_h2.groupby('brand')['top_10_popularity']
    .mean()
    .mul(100)
    .reset_index(name='pct_top10')
)

marca_stats_h2 = marca_stats.merge(pct_top10_brand, on='brand', how='left')

top_populares = marca_stats_h2.nlargest(5, 'pct_top10')

bars1 = ax1.barh(
    top_populares['brand'],
    top_populares['pct_top10'],
    color='#3498db',
    edgecolor='black'
)
ax1.set_xlabel('% Productos en Top 10% Popularidad', fontweight='bold')
ax1.set_title('Top 5 Marcas MÁS POPULARES', fontweight='bold', fontsize=13)

for bar, row in zip(bars1, top_populares.itertuples()):
    width = bar.get_width()
    ax1.text(
        width, bar.get_y() + bar.get_height()/2.,
        f'{width:.1f}% | Rating: {row.rating_promedio:.2f}',
        ha='left', va='center', fontsize=10, fontweight='bold'
    )

top_valoradas = marca_stats_h2.nlargest(5, 'rating_promedio')

bars2 = ax2.barh(
    top_valoradas['brand'],
    top_valoradas['rating_promedio'],
    color='#e74c3c',
    edgecolor='black'
)
ax2.set_xlabel('Rating Promedio', fontweight='bold')
ax2.set_title('Top 5 Marcas MEJOR VALORADAS', fontweight='bold', fontsize=13)
ax2.set_xlim(4.0, 5.0)

for bar, row in zip(bars2, top_valoradas.itertuples()):
    width = bar.get_width()
    ax2.text(
        width, bar.get_y() + bar.get_height()/2.,
        f'{width:.2f} | {row.pct_top10:.1f}% top 10',
        ha='left', va='center', fontsize=10, fontweight='bold'
    )

plt.tight_layout()
plt.savefig(h2_dir / "h2_top_marcas_comparacion.png", dpi=300, bbox_inches='tight')
print(f"✅ h2_top_marcas_comparacion.png → {h2_dir}")
plt.close()

# ============================================================
# H3: GRÁFICOS (Guardar en carpeta H3)
# ============================================================

print("\n📊 Generando gráficos para H3...")

df_h3['tiene_descuento'] = (df_h3['discount_pct'] > 0) | (df_h3['price_original'] > df_h3['price'])

# ===== H3.1: Rating y Opiniones - Con vs Sin Descuento (Global) =====
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

con_desc = df_h3[df_h3['tiene_descuento'] == True]
sin_desc = df_h3[df_h3['tiene_descuento'] == False]

data_rating = pd.DataFrame({
    'Grupo': ['Con Descuento', 'Sin Descuento'],
    'Rating': [con_desc['rating'].mean(), sin_desc['rating'].mean()],
    'N': [len(con_desc), len(sin_desc)]
})

bars1 = ax1.bar(data_rating['Grupo'], data_rating['Rating'], 
                color=['#e74c3c', '#95a5a6'], edgecolor='black', linewidth=1.5)

for bar, row in zip(bars1, data_rating.itertuples()):
    height = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2., height,
             f'{row.Rating:.3f}\n(n={row.N})',
             ha='center', va='bottom', fontweight='bold')

ax1.set_ylabel('Rating Promedio', fontweight='bold')
ax1.set_title('Rating: Con vs Sin Descuento', fontweight='bold')
ax1.set_ylim(4.0, 5.0)

data_opiniones = pd.DataFrame({
    'Grupo': ['Con Descuento', 'Sin Descuento'],
    'Opiniones': [con_desc['review_count'].mean(), sin_desc['review_count'].mean()],
})

bars2 = ax2.bar(data_opiniones['Grupo'], data_opiniones['Opiniones'], 
                color=['#3498db', '#95a5a6'], edgecolor='black', linewidth=1.5)

for bar, row in zip(bars2, data_opiniones.itertuples()):
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height,
             f'{int(row.Opiniones)}',
             ha='center', va='bottom', fontweight='bold')

ax2.set_ylabel('Opiniones Promedio', fontweight='bold')
ax2.set_title('Popularidad: Con vs Sin Descuento', fontweight='bold')

plt.tight_layout()
plt.savefig(h3_dir / "h3_descuento_global.png", dpi=300, bbox_inches='tight')
print(f"✅ h3_descuento_global.png → {h3_dir}")
plt.close()

# ===== H3.2: Rating por Gama de Precio - Con vs Sin Descuento =====
fig, ax = plt.subplots(figsize=(12, 7))

gamas = ['baja', 'media', 'alta']
x = np.arange(len(gamas))
width = 0.35

ratings_con_desc = []
ratings_sin_desc = []

for gama in gamas:
    gama_data = df_h3[df_h3['price_tier'] == gama]
    con_desc_gama = gama_data[gama_data['tiene_descuento'] == True]['rating'].mean()
    sin_desc_gama = gama_data[gama_data['tiene_descuento'] == False]['rating'].mean()
    ratings_con_desc.append(con_desc_gama)
    ratings_sin_desc.append(sin_desc_gama)

bars1 = ax.bar(x - width/2, ratings_con_desc, width, label='Con Descuento', 
               color='#e74c3c', edgecolor='black')
bars2 = ax.bar(x + width/2, ratings_sin_desc, width, label='Sin Descuento', 
               color='#2ecc71', edgecolor='black')

for bars in [bars1, bars2]:
    for bar in bars:
        height = bar.get_height()
        if not np.isnan(height):
            ax.text(bar.get_x() + bar.get_width()/2., height,
                    f'{height:.3f}',
                    ha='center', va='bottom', fontsize=10, fontweight='bold')

ax.set_xlabel('Gama de Precio', fontweight='bold')
ax.set_ylabel('Rating Promedio', fontweight='bold')
ax.set_title('H3: Rating por Gama - Con vs Sin Descuento', 
             fontweight='bold', fontsize=14)
ax.set_xticks(x)
ax.set_xticklabels([g.upper() for g in gamas])
ax.legend()
ax.set_ylim(4.0, 5.0)

plt.tight_layout()
plt.savefig(h3_dir / "h3_rating_por_gama.png", dpi=300, bbox_inches='tight')
print(f"✅ h3_rating_por_gama.png → {h3_dir}")
plt.close()

# ===== H3.3: % Top 10% con Descuento =====
fig, ax = plt.subplots(figsize=(10, 6))

top10_h3 = df_h3[df_h3['top_10_popularity'] == True]
resto_h3 = df_h3[df_h3['top_10_popularity'] == False]

data_top10_desc = pd.DataFrame({
    'Grupo': ['Top 10%\nPopularidad', 'Resto'],
    'Con_Descuento': [
        top10_h3['tiene_descuento'].sum() / len(top10_h3) * 100,
        resto_h3['tiene_descuento'].sum() / len(resto_h3) * 100
    ]
})

bars = ax.bar(data_top10_desc['Grupo'], data_top10_desc['Con_Descuento'], 
              color=['#e67e22', '#95a5a6'], edgecolor='black', linewidth=1.5)

for bar, row in zip(bars, data_top10_desc.itertuples()):
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height,
            f'{row.Con_Descuento:.1f}%',
            ha='center', va='bottom', fontweight='bold', fontsize=12)

ax.set_ylabel('% Productos con Descuento', fontweight='bold')
ax.set_title('H3: ¿Los productos populares tienen descuentos?', 
             fontweight='bold', fontsize=14)
ax.set_ylim(0, 100)
ax.axhline(y=50, color='red', linestyle='--', alpha=0.5, label='50%')
ax.legend()

plt.tight_layout()
plt.savefig(h3_dir / "h3_top10_descuentos.png", dpi=300, bbox_inches='tight')
print(f"✅ h3_top10_descuentos.png → {h3_dir}")
plt.close()

print("\n" + "="*60)
print("✅ TODOS LOS GRÁFICOS GENERADOS EN SUS CARPETAS")
print("="*60)
print(f"📁 H1 (3 gráficos): {h1_dir}")
print(f"📁 H2 (2 gráficos): {h2_dir}")
print(f"📁 H3 (3 gráficos): {h3_dir}")
print("\nTOTAL: 8 gráficos en alta resolución (300 DPI)")
```
