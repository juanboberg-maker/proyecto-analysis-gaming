```python
df_clean = df_gaming[['id', 'title', 'brand', 'price', 'price_text', 'rating', 'review_count', 'availability', 'category_name', 'category_id', 'subcategory']].copy()
```

```python
df_clean['brand'] = df_clean['brand'].str.strip().str.upper()
df_clean['title'] = df_clean['title'].str.strip()
df_clean['availability'] = df_clean['availability'].fillna('Unknown')
df_clean['review_count'] = df_clean['review_count'].fillna(0).astype(int)
df_clean['price'] = df_clean['price'].fillna(df_clean['price'].median())
df_clean['price_text'] = df_clean['price_text'].fillna('$' + df_clean['price'].astype(str))
```
